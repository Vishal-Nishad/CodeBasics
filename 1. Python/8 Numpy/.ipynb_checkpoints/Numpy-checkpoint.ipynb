## Less Time

```python
x=[1,2,3]
y=[4,5,6]
for x,y in zip(x,y):
    print(x+y)
```

```python
import time
siz=10000
l1=list(range(siz))
l2=list(range(siz))
start_time=time.time()
l3=[l1+l2 for x,y in zip(l1,l2)]
end_time=time.time()
print("python list took : ", end_time-start_time)
```

```python
siz=10000
l1=np.arange(siz)
l2=np.arange(siz)

start_t=time.time()
l3=l1+l2
end_t=time.time()
print("numpy took: ", end_t-start_t)
```

## Matrix Operations

```python
q1=np.array([[200,220,250],
             [150,180,210],
             [300,330,360]
            ])

q2=np.array([[190,231,259],
             [155,192,222],
             [310,340,375]
            ])
q1+q2
```

```python
q2-q1
```

```python
print("Percentage increase in next quarter.")
(q2-q1)*100/q1
```

```python
prices=np.array([[4,23,6],
                 [6,2,343],
                 [8,6,44]
                ])
```

```python
q1_rev=prices*q1
print(q1_rev)
```

```python
q1_dis=q1_rev*0.2
print(q1_dis)
```

```python
net_q1_rev=q1_rev-q1_dis
print(net_q1_rev)
```

```python
np.sum(net_q1_rev)
```

## Vstack , Hstack

```python
# Customer ID, Name
c = np.array([
    [101, 'Mira'],
    [102, 'Abdul'],
    [103, 'Andrea']
])
# Customer Id, Purchase Amount, Purchase Date
d = np.array([
    [101, 250.50, '2023-08-01'],
    [102, 150.00, '2023-08-02'],
    [103, 300.75, '2023-08-01']
])
```

```python
# concatenating metrices
np.hstack((c,d))              # horizontally
```

```python
e = np.array([
    [104, 'Venkat'],
    [105, 'John'],
    [106, 'Kathy'],
])
np.vstack((c,e))
```

```python
# vertically
#np.vstack((c,e))
np.vstack((c,d))
```

## vsplit , hsplit

```python
transactions = np.array([
    [101, 'Mohan', 250.50, '2023-08-01'],
    [102, 'Bob', 150.00, '2023-08-02'],
    [103, 'Fatima', 300.75, '2023-08-01'],
    [104, 'David', 400.20, '2023-08-03'],
    [105, 'Aryan', 330.1, '2023-08-04'],
])
```

```python
a,b=np.hsplit(transactions,[2])
```

```python
a
```

```python
b
```

```python
c,d=np.vsplit(transactions,[3])
```

```python
c
```

```python
d
```
