```python
# file handling converting csv file into simple dictionaries with player name and their scores
player_scores={}
with open("scores.csv") as f:
    for line in f:
        player,_,scores= line.split(",")

        scores=int(scores.strip())  # using .strip() to remove unwanted \n ie next line
        if player in player_scores:
            player_scores[player].append(scores)
        else:
            player_scores[player]=[scores]
print(player_scores)
```

```python
for player, scores_list in player_scores.items():
    min_scores=min(scores_list)
    max_scores=max(scores_list)
    average=sum(scores_list)/len(scores_list)
    print(f"{player}: \nmin_score: {min_scores}\nmax_scores: {max_scores}\naverage: {average}\n")
    
```
