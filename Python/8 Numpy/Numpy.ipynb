## Less Time

```python
x=[1,2,3]
y=[4,5,6]
for x,y in zip(x,y):
    print(x+y)
```

```python
import time
siz=10000
l1=list(range(siz))
l2=list(range(siz))
start_time=time.time()
l3=[l1+l2 for x,y in zip(l1,l2)]
end_time=time.time()
print("python list took : ", end_time-start_time)
```

```python
siz=10000
l1=np.arange(siz)
l2=np.arange(siz)

start_t=time.time()
l3=l1+l2
end_t=time.time()
print("numpy took: ", end_t-start_t)
```
