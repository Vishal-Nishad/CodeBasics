```python
import pandas as pd
df=pd.read_excel('source-code/2_pandas_basics_code/Matplotlib_seaborn_resources/linechart.xlsx')
df
```

```python
from matplotlib import pyplot as plt
```

```python
plt.figure(figsize=(12,4))  ## use to adjust the graph or chart actual size according to screen
plt.plot(df['Quarter'],df['Fridge'],color='black',label='Fridge')
plt.plot(df['Quarter'],df['Dishwasher'] ,color='blue',label='Dishwasher')
plt.plot(df['Quarter'],df['Washing Machine'],color='pink',label='Washing Machine')

plt.title("Product Sales")
plt.ylabel('Revenue')
plt.xlabel('Financial Quarter')

plt.legend()
plt.show()
```

## Pie chart

```python
df
```

```python
total_sales=df[['Fridge','Dishwasher','Washing Machine']].sum()
total_sales
```

```python
total_sales.index
```

```python
plt.pie(total_sales,labels=total_sales.index , autopct="%1.1f%%",explode=(0.1,0,0.2),startangle=90 ,counterclock=False)
plt.show()
```

## Bar Chart

```python
df.plot(kind='bar',x='Quarter')
plt.xticks(rotation=45)
```

## Histogram

```python
import seaborn as sns
sns.histplot(data=df,x="Exam_Score",kde=True)
plt.ylabel("Count")
plt.xlabel("Score")
plt.title("Exam_Score")
plt.show()
```
