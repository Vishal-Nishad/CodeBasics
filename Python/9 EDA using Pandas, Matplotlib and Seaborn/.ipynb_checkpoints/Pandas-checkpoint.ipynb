```python
import csv
with open("source-code/2_pandas_basics_code/movies.csv","r") as f:
    data=list(csv.reader(f))
    header= data[0]
    data=data[1:]

def calculate_movies_stats(data,industry=None):
    ratings=[]
    for row in data:
        if row[3]!='NULL' and (not industry or row[1]==industry):
            ratings.append(float(row[3]))
    mx_rating= max(ratings)
    min_rating= min(ratings)
    avr_rating=sum(ratings)/len(ratings)
    return mx_rating,min_rating,avr_rating
max_rat,min_rat,avr_rat = calculate_movies_stats(data)
print(f"max_ratings is : {max_rat}, min_rating is : {min_rat}, average_rating is : {avr_rat}")

max_rat,min_rat,avr_rat = calculate_movies_stats(data,industry="Bollywood")
print(f"Bollywood max_ratings is : {max_rat}, min_rating is : {min_rat}, average_rating is : {avr_rat}")

max_rat,min_rat,avr_rat = calculate_movies_stats(data, industry="Hollywood")
print(f" Hollywood max_ratings is : {max_rat}, min_rating is : {min_rat}, average_rating is : {avr_rat}")
```

## Read, Write Excel and CSV files

```python
df=pd.read_csv("stock_data.csv")
df
```

```python
df=pd.read_csv("stock_data.csv", header=1)   ## skiping 0th row and making 1st header
df
```

```python
## specifying column names
df=pd.read_csv("stock_data.csv",header =1, names=["Stock_symbol","eps", "revenue","price","people"])
df
```

```python
## filtering our data to our required no. of rows 
df=pd.read_csv("stock_data.csv",header=1,nrows=2)   ## nrows=
df
```

```python
## adding NaN value to wrong entries
df=pd.read_csv("stock_data.csv",header=1,na_values={
    'eps':[-1,'not available'],
    'revenue':[-1],
    'people':['n.a.'],
    'price' :['n.a.']
})
df
```

```python
## we can also directly specify entries without using dict of key and values,    but it is not much efficient as it will traverse each entries
df=pd.read_csv('stock_data.csv',header=1, na_values=['n.a.'])
df
```

```python
## exporting or saving a csv file to directory
df["pe"]= df['price']/df['eps']

df.to_csv('pe1.csv', index=False)  ## index=False , meanse removing index column from csv file
```

```python
pd_movies=pd.read_excel("movies_db.xlsx" ,"movies")
pd_movies.head(2)
```

```python
pd_financials=pd.read_excel("movies_db.xlsx" ,"financials")
pd_financials.head(5)
```

```python
def curr_converter(currency):
    if currency =="$$" or currency =="Dollars":
        return "USD"
    return currency
pd_financials=pd.read_excel("movies_db.xlsx" ,"financials",converters={
    "currency":curr_converter
})
pd_financials.head(5)
```

```python
## merging these two df
df_merged=pd.merge(pd_movies,pd_financials, on ="movie_id")
df_merged.head(3)
```

```python
## saving it to the directory
df_merged.to_excel("moovies_merged.xlsx", sheet_name="merged", index=False)
```

```python
## Creating a new df and then exporting it to excel file with sheet
df_weather=pd.DataFrame({
    'day':['1','2','3'],
    'tem':[28,32,38],
    'event':['Hot','Sunny','Cloudy']
})
df_weather
```

```python
with pd.ExcelWriter('weather.xlsx') as writer:
    df_weather.to_excel(writer,sheet_name="weather")
```
