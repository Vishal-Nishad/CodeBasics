```python
import csv
with open("source-code/2_pandas_basics_code/movies.csv","r") as f:
    data=list(csv.reader(f))
    header= data[0]
    data=data[1:]

def calculate_movies_stats(data,industry=None):
    ratings=[]
    for row in data:
        if row[3]!='NULL' and (not industry or row[1]==industry):
            ratings.append(float(row[3]))
    mx_rating= max(ratings)
    min_rating= min(ratings)
    avr_rating=sum(ratings)/len(ratings)
    return mx_rating,min_rating,avr_rating
max_rat,min_rat,avr_rat = calculate_movies_stats(data)
print(f"max_ratings is : {max_rat}, min_rating is : {min_rat}, average_rating is : {avr_rat}")

max_rat,min_rat,avr_rat = calculate_movies_stats(data,industry="Bollywood")
print(f"Bollywood max_ratings is : {max_rat}, min_rating is : {min_rat}, average_rating is : {avr_rat}")

max_rat,min_rat,avr_rat = calculate_movies_stats(data, industry="Hollywood")
print(f" Hollywood max_ratings is : {max_rat}, min_rating is : {min_rat}, average_rating is : {avr_rat}")
```

```python
def curr_converter(currency):
    if currency =="$$" or currency =="Dollars":
        return "USD"
    return currency
pd_financials=pd.read_excel("movies_db.xlsx" ,"financials",converters={
    "currency":curr_converter
})
pd_financials.head(5)
```

```python
When you pass curr_converter in the converters dictionary, you're not calling the function yourself—you’re simply providing a reference to the
function. The read_excel function will then call this function automatically on each value in the "currency" column.

Explanation
Function Reference vs. Function Call:

Function Call: Writing curr_converter(currency) calls the function immediately with the argument currency.

Function Reference: Writing just curr_converter passes the function itself, so that it can be called later.


How converters Works in read_excel:
The converters parameter in pd.read_excel expects a dictionary where keys are column names and values are functions.
For every cell in the specified column, read_excel will call the provided function with the cell’s value.
This means you want to pass the function itself (i.e., a reference) so that read_excel can use it for each value.

def curr_converter(currency):
    if currency == "$$" or currency == "Dollars":
        return "USD"
    return currency

pd_financials = pd.read_excel("movies_db.xlsx", "financials", converters={
    "currency": curr_converter  # Here, we pass the function reference, not calling it.
})
pd_financials.head(5)

When pd.read_excel reads the "currency" column, it does something like:
for value in currency_column:
    new_value = curr_converter(value)  # The function is called here internally.
```

```python
## merging these two df
df_merged=pd.merge(pd_movies,pd_financials, on ="movie_id")
df_merged.head(3)
```

## Group by

```python
import pandas as pd
```

```python
df=pd.read_csv("source-code/2_pandas_basics_code/6_group_by/weather_by_cities.csv")
df
```

```python
g=df.groupby('city')
```

```python
for city,data in g:
    print('city',city)
    print('data',data.temperature.max())
```

```python
print(g.max())
#print(g.mean())
g.min()
```

```python
g.size()
```

```python
g.count()
```

```python
## Grouping things according to our own made functions
def grouper(df,idx,col):
    if 80<=df[col].loc[idx]<=90:
        return '80-90'
    elif 50<=df[col].loc[idx]<80:
        return '50-79'
    else:
        return "Others"
g=df.groupby(lambda idx: grouper(df,idx,'temperature'))
g
```

```python
for key,d in g:
    print('Key',key)
    print('data',d)
```
