```python
import csv
with open("source-code/2_pandas_basics_code/movies.csv","r") as f:
    data=list(csv.reader(f))
    header= data[0]
    data=data[1:]

def calculate_movies_stats(data,industry=None):
    ratings=[]
    for row in data:
        if row[3]!='NULL' and (not industry or row[1]==industry):
            ratings.append(float(row[3]))
    mx_rating= max(ratings)
    min_rating= min(ratings)
    avr_rating=sum(ratings)/len(ratings)
    return mx_rating,min_rating,avr_rating
max_rat,min_rat,avr_rat = calculate_movies_stats(data)
print(f"max_ratings is : {max_rat}, min_rating is : {min_rat}, average_rating is : {avr_rat}")

max_rat,min_rat,avr_rat = calculate_movies_stats(data,industry="Bollywood")
print(f"Bollywood max_ratings is : {max_rat}, min_rating is : {min_rat}, average_rating is : {avr_rat}")

max_rat,min_rat,avr_rat = calculate_movies_stats(data, industry="Hollywood")
print(f" Hollywood max_ratings is : {max_rat}, min_rating is : {min_rat}, average_rating is : {avr_rat}")
```
