```python
import pandas as pd
df=pd.read_excel('source-code/2_pandas_basics_code/Matplotlib_seaborn_resources/linechart.xlsx')
df
```

```python
from matplotlib import pyplot as plt
```

```python
plt.figure(figsize=(12,4))  ## use to adjust the graph or chart actual size according to screen
plt.plot(df['Quarter'],df['Fridge'],color='black',label='Fridge')
plt.plot(df['Quarter'],df['Dishwasher'] ,color='blue',label='Dishwasher')
plt.plot(df['Quarter'],df['Washing Machine'],color='pink',label='Washing Machine')

plt.title("Product Sales")
plt.ylabel('Revenue')
plt.xlabel('Financial Quarter')

plt.legend()
plt.show()
```
