## error handling in JSON

```python
# Malformed JSON string
json_string = '''
{
    "name": "Krishna",
    "age": 30
    "is_employee": true
}
'''
```

```python
json.loads(json_string)
```

```python
## to avoid this kind of error we should always use try and catch method
try:
    json.loads(json_string)
except json.JSONDecodeError as e:
    print(f"failed to parse JSON. {e}")
```

## 12.2 Generators and Iterators

```python
prices=[]
with open('pricing.txt','r') as f:
    for line in f:
        ticker,country,price,date=line.split('|')
        prices.append({
            'Ticker':ticker,
            'Country':country,
            'Price':float(price),
            'Date':date.strip()
        })
prices[:5]
```

## Implementing custom iterator class

```python
class RemotControl:
    def __init__(self):
        self.channels=['vishal','rahul','avi']
        self.index=-1
    def __iter__(self):
        return self
    def __next__(self):
        self.index+=1
        if self.index==len(self.channels):
            raise StopIteration
        return self.channels[self.index]
   
rc=RemotControl()
for i in rc:
    print(i)

    
```

```python
## yield statement
def remote_control():
    yield 'vishal'
    yield 'avi'
    yield 'rahul'

rc=remote_control()
for i in rc:
    print(i)
```

```python
rc=remote_control()
next(rc)
```

```python
next(rc)
```

**Reading multiple files by storing it in list**

```python
def get_company_data():
    data = []
    for i in range(3):
        with open(f"stock_{i+1}.txt","r") as f:
            content = f.read()
            data.append(content)
    return data  
get_company_data()
```

**Now doing the same thing with the help of Generators -- yield statement , in this python will not create a separate memory or list , it will just give the present element and then move it to the python garbage collector**

```python
def get_company_data():
    for i in range(3):
        with open(f"stock_{i+1}.txt","r") as f:
            content = f.read()
            yield content
```

```python
print(get_company_data())
```

```python
for company_data in get_company_data():
    print(company_data)
```

```python
## Generating fibonacci series with the help of Generators
def fib():
    a,b=0,1
    while True:
        yield a
        a,b=b,a+b
fib_gen=fib()
for _ in range(10):
    print(next(fib_gen))
```

## Decorators

```python
def get_square(numbers):
    result=[]
    for n in numbers:
        result.append(n*n)
    return result

def get_cube(numbers):
    result=[]
    for n in numbers:
        result.append(n*n*n)
    return result
n=range(10)
get_square(n)
get_cube(n)
```

```python
import time
def get_square(n):
    result=[]
    start=time.time()
    for i in n:
        result.append(i*i)
    end=time.time()
    print(f"time took:{(end-start)*1000}ms")
    return result

_=get_square(range(100000))
```

```python
def get_cube(n):
    result=[]
    start=time.time()
    for i in n:
        result.append(i*i*i)
    end=time.time()
    print(f"time took:{(end-start)*1000}ms")
    return result

_=get_cube(range(100000))
```
