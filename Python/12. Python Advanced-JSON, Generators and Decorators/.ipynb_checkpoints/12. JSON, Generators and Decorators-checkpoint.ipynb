## JSON

**Converting JSON String to Dict**

```python
# JSON string
json_string = '''
{
    "name": "Maulik",
    "age": 30,
    "is_employee": true,
    "skills": ["Python", "Data Analysis", "Machine Learning"],
    "address": {
        "street": "123 Main St",
        "city": "Wonderland"
    }
}
'''
```

```python
type(json_string)
```

**## decoding this json string into python object , Dict**

```python
import json
## decoding this json string into python object , Dict
record=json.loads(json_string)
record
```

```python
type(record)
```

```python
record['age'],record['skills'],record['address']
```

**encoding python object into json string ,,i.e converting python dict to json string**

```python
data = {
    "name": "Bob",
    "age": 25,
    "is_employee": False,
    "skills": ["JavaScript", "Web Development"],
    "address": {
        "street": "456 Elm St",
        "city": "Springfield"
    }
}
type(data)
```

```python
data_str=json.dumps(data)
print(type(data_str))
data_str
```

```python
## writing or saving json file to our file directory
with open('out.json','w') as f:
    json.dump(data,f,indent=4)
```

```python
## loading a json file from directory
with open('4_json/products.json','r') as file:
    product=json.load(file)
print(type(product))
product   
```

```python
product[0]['name']
```

**Reading data from json string**

```python
json_string = '''
{
    "books": [
        {"title": "The Magic Of Thinking Big", "author": "David J. Schwartz", "published": 1959},
        {"title": "Chanakya Neeti", "author": "B.K. Chaturvedi", "published": 2009},
        {"title": "How to Win Friends and Influence People", "author": "Dale Karnegie", "published": 1936}
    ],
    "status_code": 200
}
'''
```

```python
### converting it into python object
data=json.loads(json_string)
data
```

```python
books=data['books']
books
```

## important

```python
## now just extracting only title and author with the help of list comprehension
book_list=[{'title':book['title'],'author':book['author']}for book in books]
book_list
```

## error handling in JSON

```python
# Malformed JSON string
json_string = '''
{
    "name": "Krishna",
    "age": 30
    "is_employee": true
}
'''
```

```python
json.loads(json_string)
```

```python
## to avoid this kind of error we should always use try and catch method
try:
    json.loads(json_string)
except json.JSONDecodeError as e:
    print(f"failed to parse JSON. {e}")
```
