## 9.9 Visualization : Age, Gender, Location

```python
# 1. first creating a column with age group type i.e (49-65) , (26-48),(18-25)

# define the bin edges and labels
bin_edges=[17,25,48,65]
bin_labels=["18-25","26-48","49-65"]

# Use the cut function to bin and label the age column
df_cust["age_group"]=pd.cut(df_cust["age"],bins=bin_edges,labels=bin_labels)
df_cust.head()
```

```python
# count for plotting pie chart
age_group_counts=df_cust["age_group"].value_counts(normalize=True)*100       # normalize=True -- for percentage no
age_group_counts
```

```python
plt.pie(age_group_counts,labels=age_group_counts.index,autopct="%1.1f%%",
       explode=(0.1,0,0),
        startangle=90
       )
plt.title("Distribution of Age Groups")
plt.show()
```
