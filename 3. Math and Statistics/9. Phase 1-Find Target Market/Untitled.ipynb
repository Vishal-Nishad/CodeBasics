```python
# to get overview of relation between credit limit and credit score , plot a scatter graph
plt.figure(figsize=(20,5))
plt.scatter(df_cs_clean_1["credit_limit"],df_cs_clean_1["credit_score"])
plt.title("Credit score vs credit limit")
plt.xlabel("Credit limit")
plt.ylabel("Credit score")

plt.xticks(range(0,90001,5000))
plt.grid(True)
plt.legend()
plt.show()
```

```python
# now creating a credit score range same as we created for age group range
bin_ranges=[300,450,500,550,600,650,700,750,800]
bin_labels=[f"{start}-{end-1}"for start, end in zip(bin_ranges,bin_ranges[1:])]
df_cs_clean_1["credit_score_range"]=pd.cut(df_cs_clean_1["credit_score"],bins=bin_ranges,labels=bin_labels,include_lowest=True,right=False)
df_cs_clean_1.head()
```

```python
df_cs_clean_1[df_cs_clean_1["credit_score_range"]=="700-749"]
# here we can see most data are with higher credit limit but credit limit not only depend on the credit score 
# it also depend on some other factors like income etc
```

```python
# so for substitute in our credit limit null rows we use mode of credit limit column according
# credit score range data
mode_df=df_cs_clean_1.groupby("credit_score_range")["credit_limit"].agg(lambda x: x.mode().iloc[0]).reset_index()
mode_df
```

```python
# now we merge these both df, i.e mode_df and df_df_cs_clean_1

# as there 2 credit limit col by merging so we use suffixes , keep first col as it is second as _mode
df_cs_clean_2=pd.merge(df_cs_clean_1, mode_df, on="credit_score_range",suffixes=("","_mode"))
df_cs_clean_2.sample(3)
```

```python
df_cs_clean_2[df_cs_clean_2["credit_limit"].isnull()].sample(3)
```

```python
## now filling the null value  of original credit limit column with credit limit mode

#df_cs_clean_3=df_cs_clean_2 # this create shallow copy- the underlying data remains same , so if we remove 

# null value from df 2 then the updation will reflect in df 3 also
df_cs_clean_3=df_cs_clean_2.copy() ## here we are doing deep copy ,, any updation in df2 will not reflect here
df_cs_clean_2["credit_limit"].fillna(df_cs_clean_2["credit_limit_mode"],inplace=True)
                                     
                                    
```
