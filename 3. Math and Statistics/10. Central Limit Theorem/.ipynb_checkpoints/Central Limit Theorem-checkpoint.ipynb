```python
# Random Sampling & Sample Bias
# the law of large numbers(LLN)
# Central Limit Theorem, Sampling Distribution
```

```python
import pandas as pd
from matplotlib import pyplot as plot
import seaborn as sns
```

# analysis to prove central limit theorem

```python
df=pd.read_csv("chapter9_assets/4_clt_solar_panels_coding/bill_survey.csv")
df.head()
```

```python
df["City"].nunique()  # nunique() is for see no. of unique records
```

```python
sns.histplot(df["Bill"],kde=True)
# distribution for all data
```

```python
# for specific city
# for mumbai
sns.histplot(df[df["City"]=="Mumbai"]["Bill"],kde=True)
```

```python
# for surat
sns.histplot(df[df["City"]=="Surat"]["Bill"],kde=True)
```

```python
# so we take each city mean value of bill to plot sample distribution

for city,sub_df in df.groupby("City"):
    print(city)
    print(sub_df) # print(city) → Displays the name of the city (group).
                    # print(sub_df) → Displays all the rows in df that belong to that city
```

```python
sample_means=[]
for city,subdf in df.groupby("City"):
    print(city, subdf["Bill"].mean()) # printing mean city wise
```

```python
sample_means=[]
for city,subdf in df.groupby("City"):
    sample_means.append(subdf["Bill"].mean())
sample_means[:5]
```

```python
len(sample_means)
```

```python
# sample distribution for all samples city mean values
sns.histplot(sample_means,kde=True)
# hence it is a normal distribution 
```
