```python
sample_means=[]
for city,subdf in df.groupby("City"):
    print(city, subdf["Bill"].mean()) # printing mean city wise
```

```python
sample_means=[]
for city,subdf in df.groupby("City"):
    sample_means.append(subdf["Bill"].mean())
sample_means[:5]
```

```python
len(sample_means)
```

```python
# sample distribution for all samples city mean values
sns.histplot(sample_means,kde=True)
# hence it is a normal distribution 
```

# 10.10 , Confidence Interval: implementation , Estimates car miles

```python
df_car=pd.read_csv("chapter9_assets/8_Confidence Interval Estimate Car Miles/miles.csv")
df_car.head()
```

```python
df_car.shape
```

```python
plt.figure(figsize=(12,4))
plt.plot(df_car["miles"])
```

```python
sns.histplot(df_car["miles"],kde=True)
```

```python
# mean of our whole dataset 
population_mean=df_car["miles"].mean()
population_std=df_car["miles"].std()
population_mean,population_std
```

```python
sample_size=50
sample=df_car.sample(sample_size,random_state=100)  # every time it will give same result
sample.head()
```

```python
sample_mean=sample["miles"].mean()
sample_std=sample["miles"].std()
sample_mean,sample_std
```

```python
import scipy.stats as st
import numpy as np
# calculating z_score for percentage of confidence level

 # For a 95% confidence level, the Z-score is ±1.96 because it leaves 2.5% on each side of the normal distribution.
 # If we wanted a 90% confidence level, we’d look at 95% cumulative probability (since 5% would be in the right tail).


z_score_95_confidence=st.norm.ppf(0.975)   
z_score_95_confidence
```

```python
margin_of_error=z_score_95_confidence*(population_std/np.sqrt(sample_size))
margin_of_error
```

```python
# calculating confidence interval/range
lower=sample_mean-margin_of_error
upper=sample_mean+margin_of_error
lower,upper
```
