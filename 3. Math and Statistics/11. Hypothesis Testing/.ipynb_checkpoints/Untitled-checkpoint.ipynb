# Housing inflation test : Z_test , Rejection Region

```python
import pandas as pd
import numpy as np
```

```python
#H_o=inflation of house is 10
#H_a=that it is greater than 10
```

```python
# df of sample data contain 100 items
df=pd.read_csv("chapter10_assets/3_z_test_housing_inflation/house_price_increase.csv")
df.head()
```

```python
df.shape
```

```python
# Given data, i.e population_mean, population_std_dev
population_mean=10
population_std=4
```

```python
sample_mean=df["price_increase_pct"].mean()
sample_mean
```

```python
# calculating standard error ,, i.e population_std/sqrt(sample size)
```

```python
standard_error=population_std/np.sqrt(df.shape[0])
standard_error
```

```python
z_score=(sample_mean-population_mean)/standard_error
z_score
```

```python
alpha = 0.05 # significance level i.e alpha 
```

```python
from scipy import stats
z_critical=stats.norm.ppf(1-alpha)
z_critical
```

```python
z_score,z_critical
```

-Since z_score > z_critical , i.e fall in rejection region , hence we will reject the 
-Null Hypothesis. This means alternate hypothesis becomes true which means the inflation-of 
-house is greater than 10%

# Z_test using P_value

```python
# cumulative percentage for z_score 2.5
stats.norm.cdf(z_score)
```

```python
# caclulating p_value from z_score percentage
p_value=1-stats.norm.cdf(z_score)
p_value
```

```python
p_value,alpha
```

-Since P_value (0.006) is less than alpha (i.e significance level 0.05) , hence we reject the null hypothesis. This means there is statistically significant evidence to support the claim that the inflation rate in house prices is higher than the reported 10%
